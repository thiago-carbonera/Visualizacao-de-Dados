```python
import pandas as pd
import plotly.express as px

# 1. Leitura do dataset
df = pd.read_csv("../data/chicago_crime_sample.csv")

# 2. Garantindo que os anos estão ordenados
df = df.sort_values('year')

# 3. Agrupamento
crime_by_year = df.groupby(['year', 'primary_type']).size().reset_index(name='Count')

# 4. Filtragem, Pegando apenas os Top 10 crimes globais
top_10_crimes = df['primary_type'].value_counts().nlargest(10).index
df_animado = crime_by_year[crime_by_year['primary_type'].isin(top_10_crimes)]

# 5. Gráfico de Barras Animado
fig = px.bar(
    df_animado,
    x="Count",
    y="primary_type",
    animation_frame="year",
    animation_group="primary_type", 
    orientation='h',
    color="primary_type",
    title="Evolução Temporal dos Top 10 Crimes em Chicago (2001-Presente)",
    range_x=[0, df_animado['Count'].max() + 500], 
    labels={'primary_type': 'Tipo de Crime', 'Count': 'Quantidade'}
)

# 6. Ajustes de Layout e Velocidade
fig.update_layout(
    yaxis={'categoryorder':'total ascending'},
    template="plotly_white",
    showlegend=False
)

fig.layout.updatemenus[0].buttons[0].args[1]['frame']['duration'] = 500

# Exibe o gráfico interativo
fig.show()

# 7. Salva como HTML
fig.write_html("../outputs/animacao_top15_crimes.html")
```

```python
# 1. Limpeza de dados geográficos 
df_mapa = df.dropna(subset=['latitude', 'longitude'])

# 2. Cria o mapa de calor consolidado
fig_mapa = px.density_mapbox(
    df_mapa, 
    lat='latitude', 
    lon='longitude', 
    radius=1,  # Radiu de 1 para garantir a precisão das ruas e bairros
    center=dict(lat=41.8781, lon=-87.6298), 
    zoom=11, 
    mapbox_style="carto-positron", 
    title="Mapa de Calor: Concentração Geral de Crimes em Chicago"
)

# 3. Exibe o mapa interativo na tela
fig_mapa.show()

# 4. Salva o gráfico como uma página HTML
fig_mapa.write_html("../outputs/heatmap_crimes.html")
```
