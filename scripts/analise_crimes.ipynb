```python
import pandas as pd
import plotly.express as px

# 1. Leitura do dataset
df = pd.read_csv("../data/chicago_crime_sample.csv")

# 2. Pré-processament, Converte a coluna 'date' para datetime
df['date'] = pd.to_datetime(df['date'], errors='coerce')

# 3. Contagem dos crimes (Top 15)
crime_counts = df["primary_type"].value_counts().reset_index()
crime_counts.columns = ["Crime Type", "Count"]
top_crimes = crime_counts.head(15)

# 4. Gráfico de Barras Interativo
fig = px.bar(
    top_crimes,
    x="Count", 
    y="Crime Type",
    orientation='h', # Barras horizontais para facilitar a leitura
    text="Count",
    title="Top 15 Tipos de Crimes Mais Frequentes em Chicago",
    color="Count", # Adiciona uma escala de cores baseada no volume
    color_continuous_scale="Reds"
)

# 5. Ajustes de layout
fig.update_layout(
    xaxis_title="Quantidade de Ocorrências",
    yaxis_title="Tipo de Crime",
    yaxis={'categoryorder':'total ascending'}, # Garante que o maior fique no topo
    template="plotly_white"
)

# Exibe o gráfico interativo
fig.show()

# salva como PNG
fig.write_html("../outputs/top15_crimes.html")
```

```python
# 1. Limpeza de dados geográficos 
df_mapa = df.dropna(subset=['latitude', 'longitude'])

# 2. Cria o mapa de calor consolidado
fig_mapa = px.density_mapbox(
    df_mapa, 
    lat='latitude', 
    lon='longitude', 
    radius=1,  # Radiu de 1 para garantir a precisão das ruas e bairros
    center=dict(lat=41.8781, lon=-87.6298), 
    zoom=11, 
    mapbox_style="carto-positron", 
    title="Mapa de Calor: Concentração Geral de Crimes em Chicago"
)

# 3. Exibe o mapa interativo na tela
fig_mapa.show()

# 4. Salva o gráfico como uma página HTML
fig_mapa.write_html("../outputs/heatmap_crimes.html")
```
